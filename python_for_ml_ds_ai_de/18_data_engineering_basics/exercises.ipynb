## Common Pitfalls

1. **Not handling API rate limits** — get blocked or throttled.
   *Fix:* Implement token bucket or sliding window rate limiting.

2. **Using CSV for large ML datasets** — huge file sizes, slow reads.
   *Fix:* Use Parquet with snappy compression.

3. **Not retrying failed tasks** — transient failures cause pipeline failures.
   *Fix:* Configure retries with exponential backoff in Airflow.

4. **Ignoring data quality** — bad data leads to bad models.
   *Fix:* Always validate data before training or inference.

5. **Hardcoding paths and credentials** — breaks across environments.
   *Fix:* Use environment variables and configuration files.

6. **No pipeline monitoring** — silent failures go unnoticed.
   *Fix:* Add logging, metrics, and alerting to all pipelines.
