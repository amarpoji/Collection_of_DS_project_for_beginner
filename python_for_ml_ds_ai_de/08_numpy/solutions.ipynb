## Exercise 2: Reshaping and Transposing

```python
arr = np.arange(24)
print('Original:', arr, 'shape:', arr.shape)

r46 = arr.reshape(4, 6)
print('Reshape (4,6):', r46.shape)
print(r46)

r234 = arr.reshape(2, 3, 4)
print('\nReshape (2,3,4):', r234.shape)
print(r234)

transposed = r46.T
print('\nTransposed (6,4):', transposed.shape)
print(transposed)

flat = r46.flatten()
print('\nFlattened:', flat)
print('Matches original:', np.array_equal(flat, arr))
```

## Exercise 3: Performance Comparison

```python
import time
import math
import numpy as np

n = 1_000_000
py_list = [float(i) / 1000.0 for i in range(n)]
np_arr = np.arange(n, dtype=float) / 1000.0

# Python loop
start = time.time()
py_result = [math.sin(x)**2 + math.cos(x)**2 for x in py_list]
py_time = time.time() - start

# NumPy
start = time.time()
np_result = np.sin(np_arr)**2 + np.cos(np_arr)**2
np_time = time.time() - start

print('Python time:', round(py_time, 3), 'sec')
print('NumPy time:', round(np_time, 3), 'sec')
print('Speedup:', round(py_time / np_time, 1), 'x')
print('\nVerification (first 5):')
print('  Python:', [round(v, 6) for v in py_result[:5]])
print('  NumPy:', np.round(np_result[:5], 6))
```

## Exercise 4: Broadcasting

```python
np.random.seed(42)
X = np.random.randn(5, 3)

print('Original:')
print(np.round(X, 3))

mean = np.mean(X, axis=0)
std = np.std(X, axis=0)
print('\nFeature means:', np.round(mean, 3))
print('Feature stds:', np.round(std, 3))

X_std = (X - mean) / std
print('\nStandardized:')
print(np.round(X_std, 3))
print('\nVerification:')
print('  Mean after:', np.round(np.mean(X_std, axis=0), 10))
print('  Std after:', np.round(np.std(X_std, axis=0), 10))

# Add bias column
bias = np.ones((X_std.shape[0], 1))
X_with_bias = np.hstack([bias, X_std])
print('\nWith bias column:', X_with_bias.shape)
print(X_with_bias[:3])
```

## Exercise 5: Boolean Indexing

```python
np.random.seed(42)
arr = np.random.randint(0, 101, 50)

print('Original:', arr)

above_75 = arr[arr > 75]
print('\nValues > 75:', above_75, '| Count:', len(above_75))

between_30_60 = arr[(arr >= 30) & (arr <= 60)]
print('Values 30-60:', between_30_60, '| Count:', len(between_30_60))

modified = arr.copy()
modified[modified < 20] = 0
modified[modified > 80] = 80

print('\nModified:')
print('  Values < 20 set to 0')
print('  Values > 80 set to 80')
print('  Result:', modified)

count_exact_50 = np.sum(arr == 50)
print('\nCount of exact 50:', count_exact_50)
```

## Exercise 6: Aggregation

```python
np.random.seed(42)
data = np.random.randn(10, 4)

print('Per feature (axis=0):')
print('  Mean:', np.round(np.mean(data, axis=0), 3))
print('  Min:', np.round(np.min(data, axis=0), 3))
print('  Max:', np.round(np.max(data, axis=0), 3))
print('  Std:', np.round(np.std(data, axis=0), 3))

print('\nPer sample (axis=1):')
print('  Mean:', np.round(np.mean(data, axis=1), 3))
print('  Min:', np.round(np.min(data, axis=1), 3))
print('  Max:', np.round(np.max(data, axis=1), 3))
print('  Std:', np.round(np.std(data, axis=1), 3))

variances = np.var(data, axis=0)
highest_var_feat = np.argmax(variances)
print('\nFeature with highest variance:', highest_var_feat, '(variance:', round(variances[highest_var_feat], 3), ')')

sample_means = np.mean(data, axis=1)
highest_mean_sample = np.argmax(sample_means)
print('Sample with highest mean:', highest_mean_sample, '(mean:', round(sample_means[highest_mean_sample], 3), ')')
```

## Exercise 7: Linear Algebra

```python
from numpy import linalg

np.random.seed(42)
A = np.random.randn(3, 3)
B = np.random.randn(3, 3)

AB = A @ B
BA = B @ A
print('A @ B == B @ A:', np.allclose(AB, BA))  # Usually False!

det = linalg.det(A)
print('\nDet(A):', round(det, 4))

A_inv = linalg.inv(A)
print('\nA @ inv(A):')
print(np.round(A @ A_inv, 10))
print('Close to identity:', np.allclose(A @ A_inv, np.eye(3)))

eigvals, eigvecs = linalg.eig(A)
print('\nEigenvalues:', np.round(eigvals, 4))
print('Eigenvectors:')
print(np.round(eigvecs, 4))
```

## Exercise 8: Train-Test Split from Scratch

```python
def my_train_test_split(X, y, test_size=0.2, random_seed=42):
    np.random.seed(random_seed)
    n = len(X)
    n_test = int(n * test_size)
    indices = np.random.permutation(n)
    test_idx = indices[:n_test]
    train_idx = indices[n_test:]
    return X[train_idx], X[test_idx], y[train_idx], y[test_idx]

# Test
np.random.seed(42)
X = np.random.randn(100, 5)
y = np.random.randint(0, 2, 100)

X_train, X_test, y_train, y_test = my_train_test_split(X, y, test_size=0.2)

print('X_train shape:', X_train.shape)
print('X_test shape:', X_test.shape)
print('y_train shape:', y_train.shape)
print('y_test shape:', y_test.shape)
print('\nTotal:', X_train.shape[0] + X_test.shape[0], 'samples')
```

## Exercise 9: Mean Squared Error

```python
def mse(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)

def rmse(y_true, y_pred):
    return np.sqrt(mse(y_true, y_pred))

# Test with random data
np.random.seed(42)
y_true = np.random.randn(100)
y_pred = y_true + np.random.randn(100) * 0.1  # Add noise

print('MSE:', round(mse(y_true, y_pred), 6))
print('RMSE:', round(rmse(y_true, y_pred), 6))

# Manual verification
small_y = np.array([1, 2, 3])
small_pred = np.array([1.1, 1.9, 3.2])
manual = ((1-1.1)**2 + (2-1.9)**2 + (3-3.2)**2) / 3
print('\nManual MSE:', round(manual, 6))
print('Function MSE:', round(mse(small_y, small_pred), 6))
print('Match:', abs(manual - mse(small_y, small_pred)) < 1e-10)
```

## Exercise 10: One-Hot Encoding

```python
def one_hot(y, num_classes):
    return np.eye(num_classes)[y]

# Test
y = np.array([0, 2, 1, 2, 0, 1])
encoded = one_hot(y, 3)

print('Original:', y)
print('One-hot encoded:')
print(encoded)
print('Shape:', encoded.shape)

# Verify
decoded = np.argmax(encoded, axis=1)
print('\nDecoded back:', decoded)
print('Matches original:', np.array_equal(y, decoded))
```
