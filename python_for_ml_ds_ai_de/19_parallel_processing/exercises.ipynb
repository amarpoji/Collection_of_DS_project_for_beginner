## Common Pitfalls

1. **Using threads for CPU-bound tasks** — GIL prevents true parallelism.
   *Fix:* Use multiprocessing or ProcessPoolExecutor.

2. **Race conditions with shared data** — multiple threads modifying same data.
   *Fix:* Use locks (`threading.Lock`) or avoid shared state.

3. **Creating too many processes** — overhead from context switching.
   *Fix:* Limit to `cpu_count()` workers.

4. **Mixing blocking code with asyncio** — blocks the event loop.
   *Fix:* Use `asyncio.to_thread()` or `loop.run_in_executor()`.

5. **Serializing large data to processes** — pickle overhead.
   *Fix:* Use shared memory or chunk data intelligently.

6. **Over-engineering** — parallelizing code that runs in 0.1s sequentially.
   *Fix:* Profile first, then parallelize only bottlenecks.
