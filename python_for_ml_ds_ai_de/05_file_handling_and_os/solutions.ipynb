## Exercise 1: Read a CSV and Compute Statistics

```python
import csv

# Step 1: Write CSV
data = [
    {'name': 'Alice', 'height_cm': 165, 'weight_kg': 60},
    {'name': 'Bob', 'height_cm': 180, 'weight_kg': 78},
    {'name': 'Charlie', 'height_cm': 175, 'weight_kg': 72},
    {'name': 'Diana', 'height_cm': 160, 'weight_kg': 55},
    {'name': 'Eve', 'height_cm': 170, 'weight_kg': 65},
]

with open('heights.csv', 'w', newline='') as f:
    writer = csv.DictWriter(f, fieldnames=['name', 'height_cm', 'weight_kg'])
    writer.writeheader()
    writer.writerows(data)

# Step 2: Read CSV and compute averages
total_height = 0
total_weight = 0
count = 0

with open('heights.csv', 'r') as f:
    reader = csv.DictReader(f)
    for row in reader:
        total_height += float(row['height_cm'])
        total_weight += float(row['weight_kg'])
        count += 1

avg_height = total_height / count
avg_weight = total_weight / count

print('Average height:', avg_height, 'cm')
print('Average weight:', avg_weight, 'kg')
```

## Exercise 2: JSONL Dataset Reader

```python
import json

# Create samples
samples = [
    {'id': 1, 'text': 'Great product!', 'label': 'positive'},
    {'id': 2, 'text': 'Terrible service.', 'label': 'negative'},
    {'id': 3, 'text': 'Works well.', 'label': 'positive'},
    {'id': 4, 'text': 'Not good.', 'label': 'negative'},
    {'id': 5, 'text': 'Excellent quality.', 'label': 'positive'},
    {'id': 6, 'text': 'Could be better.', 'label': 'negative'},
    {'id': 7, 'text': 'Love it!', 'label': 'positive'},
    {'id': 8, 'text': 'Very disappointed.', 'label': 'negative'},
]

# Write JSONL
with open('reviews.jsonl', 'w') as f:
    for s in samples:
        f.write(json.dumps(s) + '\n')

# Read and count labels
positive_count = 0
negative_count = 0

with open('reviews.jsonl', 'r') as f:
    for line in f:
        sample = json.loads(line.strip())
        if sample['label'] == 'positive':
            positive_count += 1
        else:
            negative_count += 1

print('Positive samples:', positive_count)
print('Negative samples:', negative_count)
```

## Exercise 4: Model Artifact Manager

```python
import json
from pathlib import Path

def save_model_artifact(model_info, filepath):
    filepath = Path(filepath)
    
    # Save main artifact
    with open(filepath, 'w') as f:
        json.dump(model_info, f, indent=2)
    
    # Save companion metadata
    meta_path = filepath.with_name(filepath.stem + '_meta.json')
    metadata = {
        'model_file': str(filepath),
        'accuracy': model_info['accuracy'],
        'file_size': filepath.stat().st_size
    }
    with open(meta_path, 'w') as f:
        json.dump(metadata, f, indent=2)
    
    return filepath.stat().st_size, meta_path.stat().st_size

# Test with sample data
model_info = {
    'name': 'linear_regression',
    'accuracy': 0.89,
    'params': {'learning_rate': 0.01, 'epochs': 100},
    'timestamp': '2025-01-15T10:30:00'
}

main_size, meta_size = save_model_artifact(model_info, 'model_artifact.json')
print('Main artifact size:', main_size, 'bytes')
print('Metadata size:', meta_size, 'bytes')
```

## Exercise 5: Large File Processing

```python
# Create large file
with open('large_log.txt', 'w') as f:
    for i in range(1, 1001):
        f.write('Line ' + str(i) + ': Some log data\n')

# Process line by line (memory efficient)
count = 0
last_line = ''

with open('large_log.txt', 'r') as f:
    for line in f:
        if 'log' in line:
            count += 1
        last_line = line.strip()

print('Lines containing "log":', count)
print('Last line:', last_line)
print('Method used: iteration (for line in f) — memory efficient')
```

## Exercise 6: File Metadata Report

```python
import time
from pathlib import Path

# Create files of different sizes
files_info = [
    ('small_file.txt', 10),
    ('medium_file.txt', 100),
    ('large_file.txt', 1000),
]

for name, size in files_info:
    with open(name, 'w') as f:
        f.write('x' * size)

# Report metadata
largest = ('', 0)
print('File Metadata Report:')
print('-' * 50)

for name, _ in files_info:
    p = Path(name)
    stat = p.stat()
    mod_time = time.ctime(stat.st_mtime)
    size = stat.st_size
    print('File:', name)
    print('  Size:', size, 'bytes')
    print('  Modified:', mod_time)
    print()
    if size > largest[1]:
        largest = (name, size)

print('Largest file:', largest[0], '-', largest[1], 'bytes')
```
