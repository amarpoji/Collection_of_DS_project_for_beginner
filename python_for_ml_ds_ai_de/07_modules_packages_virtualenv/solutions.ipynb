## Exercise 1: The `__name__` Guard

```python
# Create calculator.py
calc_code = '''\
def add(a, b):\n    return a + b\n\ndef subtract(a, b):\n    return a - b\n\ndef multiply(a, b):\n    return a * b\n\ndef divide(a, b):\n    if b == 0:\n        return "Error: division by zero"\n    return a / b\n\nif __name__ == '__main__':\n    print("Testing calculator...")\n    print("add(5, 3) =", add(5, 3))\n    print("subtract(5, 3) =", subtract(5, 3))\n    print("multiply(5, 3) =", multiply(5, 3))\n    print("divide(5, 3) =", divide(5, 3))\n'''

with open('calculator.py', 'w') as f:
    f.write(calc_code)

# Import (should NOT print the test output)
import calculator

print('Imported calculator module')
print('add(5, 3) =', calculator.add(5, 3))
print('The test code did NOT run on import — __name__ guard works')
```

## Exercise 2: Create a Package

```python
from pathlib import Path

# Create package structure
base = Path('ml_utils')
(base / 'data').mkdir(parents=True, exist_ok=True)
(base / 'model').mkdir(parents=True, exist_ok=True)

# Top-level __init__.py
with open(base / '__init__.py', 'w') as f:
    f.write('from .data.io import load_csv\n')
    f.write('from .model.split import train_test_split\n')

# data/__init__.py
with open(base / 'data' / '__init__.py', 'w') as f:
    f.write('from .io import load_csv\n')

# data/io.py
with open(base / 'data' / 'io.py', 'w') as f:
    f.write('def load_csv(path):\n')
    f.write('    print("Loading CSV from:", path)\n')
    f.write('    return {"data": [1, 2, 3], "path": path}\n')

# model/__init__.py
with open(base / 'model' / '__init__.py', 'w') as f:
    f.write('from .split import train_test_split\n')

# model/split.py
with open(base / 'model' / 'split.py', 'w') as f:
    f.write('def train_test_split(data, test_size=0.2):\n')
    f.write('    print("Splitting data with test_size:", test_size)\n')
    f.write('    return {"train": [1, 2], "test": [3]}\n')

# Import and use
import sys
sys.path.insert(0, '.')

from ml_utils import load_csv, train_test_split

data = load_csv('dataset.csv')
print('Data:', data)

split = train_test_split(data, test_size=0.3)
print('Split:', split)
```

## Exercise 3: Import Styles

```python
from pathlib import Path

# Create analysis package
base = Path('analysis')
(base / 'utils').mkdir(parents=True, exist_ok=True)

# analysis/__init__.py
with open(base / '__init__.py', 'w') as f:
    f.write('from .stats import mean, std\n')

# analysis/stats.py
with open(base / 'stats.py', 'w') as f:
    f.write('def mean(data):\n')
    f.write('    return sum(data) / len(data)\n')
    f.write('def std(data):\n')
    f.write('    m = mean(data)\n')
    f.write('    return (sum((x - m)**2 for x in data) / len(data))**0.5\n')

# analysis/utils/__init__.py
with open(base / 'utils' / '__init__.py', 'w') as f:
    f.write('\n')

# analysis/utils/helpers.py (with relative import)
helpers_code = '''\
from ..stats import mean  # relative import\n\ndef normalize(data):\n    m = mean(data)\n    return [x - m for x in data]\n'''
with open(base / 'utils' / 'helpers.py', 'w') as f:
    f.write(helpers_code)

# Test imports
import sys
sys.path.insert(0, '.')

from analysis import mean
from analysis.utils.helpers import normalize

data = [1, 2, 3, 4, 5]
print('mean:', mean(data))
print('normalized:', normalize(data))
```

## Exercise 5: Module Search Path

```python
import sys
from pathlib import Path

print('Original sys.path:')
for p in sys.path[:3]:
    print(' ', p)

# Add custom directory
custom_dir = Path('my_custom_modules')
custom_dir.mkdir(exist_ok=True)
sys.path.insert(0, str(custom_dir))

# Create hello.py
with open(custom_dir / 'hello.py', 'w') as f:
    f.write('def greet():\n')
    f.write('    print("Hello from custom module!")\n')

# Import and use
import hello
hello.greet()

# Remove custom path
sys.path.remove(str(custom_dir))
print('\nRemoved custom path from sys.path')
```

## Exercise 6: Create a Console Script

```python
from pathlib import Path

# Create greeter package
base = Path('greeter')
base.mkdir(exist_ok=True)

with open(base / '__init__.py', 'w') as f:
    f.write('from .greet import say_hello\n')

greet_code = '''\
import sys\n\ndef say_hello(name):\n    print("Hello, " + name + "!")\n\ndef main():\n    if len(sys.argv) > 1:\n        say_hello(sys.argv[1])\n    else:\n        print("Usage: python greet.py <name>")\n\nif __name__ == '__main__':\n    main()\n'''

with open(base / 'greet.py', 'w') as f:
    f.write(greet_code)

# Test by calling the function directly
from greeter import say_hello
say_hello('ML Engineer')

print()
print('To run as script:')
print('  python greeter/greet.py Alice')
```
