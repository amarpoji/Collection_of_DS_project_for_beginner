## Solution 1: Handle Missing Values

```python
import pandas as pd
import numpy as np

data = {"A": [1, 2, np.nan, 4, 5], "B": [np.nan, 2, 3, np.nan, 5], "C": [1, 2, 3, 4, 5]}
df = pd.DataFrame(data)

# Method 1: Drop rows with any NaN
print("Drop rows:")
print(df.dropna())

# Method 2: Fill with mean
print("\nFill with mean:")
print(df.fillna(df.mean()))

# Method 3: Forward fill
print("\nForward fill:")
print(df.ffill())

# Method 4: Interpolate
print("\nInterpolate:")
print(df.interpolate())
```

## Solution 2: Detect and Handle Outliers

```python
np.random.seed(42)
data = np.concatenate([np.random.normal(50, 10, 95),
                       np.array([5, 8, 150, 160, 200])])
values = pd.Series(data)

Q1 = values.quantile(0.25)
Q3 = values.quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = values[(values < lower) | (values > upper)]
print("Outliers found:", len(outliers))
print("Outlier values:", outliers.tolist())

# Cap outliers
capped = values.clip(lower, upper)
print("\nCapped stats:")
print(capped.describe())
```

## Solution 3: Clean Inconsistent Text

```python
raw = pd.Series(["  data scientist  ", "ML ENGINEER", "Data Scientist",
                  "ml engineer", "DATA scientist", "  Data  Engineer  "])

# Clean: strip whitespace, title case, remove multiple spaces
import re
cleaned = raw.str.strip().str.title()
cleaned = cleaned.str.replace(r"\s+", " ", regex=True)

print("Cleaned values:")
print(cleaned)
print("\nUnique:", cleaned.nunique())
print("Value counts:")
print(cleaned.value_counts())
```

## Solution 4: Titanic EDA

```python
import seaborn as sns
titanic = sns.load_dataset('titanic')

print("Shape:", titanic.shape)
print("\nInfo:")
print(titanic.info())
print("\nMissing %:")
print(titanic.isna().mean().sort_values(ascending=False) * 100)
print("\nSurvival by sex:")
print(titanic.groupby('sex')['survived'].mean())
print("\nSurvival by class:")
print(titanic.groupby('class')['survived'].mean())
print("\nCorrelation matrix:")
print(titanic.select_dtypes(include=[np.number]).corr())
```

## Solution 5: Feature Engineering from Dates

```python
dates = pd.date_range('2023-01-01', periods=100, freq='D')
df = pd.DataFrame({'date': dates, 'sales': np.random.randint(100, 500, 100)})

df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month
df['day'] = df['date'].dt.day
df['dayofweek'] = df['date'].dt.dayofweek
df['is_weekend'] = df['dayofweek'].isin([5, 6]).astype(int)
df['quarter'] = df['date'].dt.quarter

print("Features created:")
print(df.head())
print("\nAverage sales by day of week:")
print(df.groupby('dayofweek')['sales'].mean())
print("\nWeekday vs Weekend:")
print(df.groupby('is_weekend')['sales'].agg(['mean', 'std']))
```
