## Your First Data Science Script

Let's write a complete mini data science workflow:

```python
def analyze_data() -> None:
    """Demonstrate a mini data science workflow."""
    # Create simple dataset
    data: dict = {
        "name": ["Alice", "Bob", "Charlie", "Diana"],
        "age": [25, 32, 37, 29],
        "salary": [70000, 85000, 95000, 72000]
    }
    
    df = pd.DataFrame(data)
    print("📊 Dataset:")
    print(df)
    print(f"\n📈 Summary statistics:")
    print(df.describe())
    
    # Simple visualization
    plt.figure(figsize=(8, 4))
    plt.bar(df["name"], df["salary"])
    plt.title("Salary by Employee")
    plt.ylabel("Salary ($)")
    plt.show()

analyze_data()
```
