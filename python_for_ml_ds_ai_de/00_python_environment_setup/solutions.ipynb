## Solution 6: Package Version Check

```python
import importlib

def check_package(package_name: str) -> str:
    """Check if a package is installed and return its version."""
    try:
        module = importlib.import_module(package_name)
        return f"{package_name}: {module.__version__}"
    except ImportError:
        return f"{package_name}: Not installed"

print(check_package("pandas"))
print(check_package("nonexistent_package"))

# Alternative using pkg_resources
import importlib.metadata as metadata
def check_package_v2(name: str) -> str:
    try:
        v = metadata.version(name)
        return f"{name}: {v}"
    except metadata.PackageNotFoundError:
        return f"{name}: Not installed"
print(check_package_v2("seaborn"))
```
