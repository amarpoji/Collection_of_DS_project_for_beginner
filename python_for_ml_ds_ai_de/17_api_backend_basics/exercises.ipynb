## Common Pitfalls

1. **Loading model inside endpoint function** — loads model on every request, causing extreme slowdown.
   *Fix:* Load model once with `@app.on_event('startup')` or the `lifespan` pattern.

2. **Blocking the event loop** — running CPU-bound tasks (like model inference) in async endpoints blocks all other requests.
   *Fix:* Use `run_in_threadpool` for CPU-bound work or keep endpoints synchronous.

3. **Not validating inputs** — trusting raw user input can crash your model or produce meaningless predictions.
   *Fix:* Use Pydantic models with `Field(ge=..., le=...)` constraints matching training data ranges.

4. **Hardcoding configuration** — makes deployment to different environments (dev, staging, production) difficult.
   *Fix:* Use `pydantic-settings` with `.env` files for all configurable values.

5. **Forgetting CORS** — frontend apps get blocked by browser CORS policy when calling your API from a different origin.
   *Fix:* Add `CORSMiddleware` with appropriate `allow_origins` before any routes.

6. **No error handling for model failures** — if model prediction raises an exception, the API returns a generic 500.
   *Fix:* Wrap prediction in try/except and return appropriate HTTP status codes with descriptive details.

7. **Route ordering issues** — placing `/{listing_id}` before `/search` causes `/search` to never match.
   *Fix:* Define static routes before parameterized routes.

8. **Exposing sensitive information** — returning stack traces or internal details in error responses.
   *Fix:* In production, log the full error server-side and return a sanitized error message to the client.
