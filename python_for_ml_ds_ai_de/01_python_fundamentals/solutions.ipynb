## Exercise 1: Temperature Converter

```python
f = 100
c = (f - 32) * 5/9
print(f"{f}F = {c:.1f}C")
```

## Exercise 2: List Statistics

```python
scores = [85, 92, 78, 95, 88, 72, 90]
print("Min:", min(scores), "Max:", max(scores))
print("Sum:", sum(scores), "Mean:", sum(scores)/len(scores))
above_80 = [x for x in scores if x > 80]
print("Count above 80:", len(above_80))
```

## Exercise 3: Data Cleaner

```python
raw = "Alice,25,Engineer|Bob,30,Doctor|Charlie,35,Teacher"
result = []
for entry in raw.split("|"):
    f = entry.split(",")
    result.append({"name": f[0], "age": int(f[1]), "job": f[2]})
print(result)
```

## Exercise 4: Word Frequency

```python
text = "the quick brown fox jumps over the lazy dog. The dog barks!"
import re
words = re.findall(r"\w+", text.lower())
freq = {}
for w in words:
    freq[w] = freq.get(w, 0) + 1
print(freq)
```

## Exercise 5: Set Operations

```python
A, B = [101,102,103,104,105], [103,104,106,107,108]
sA, sB = set(A), set(B)
print("Both:", sA & sB)
print("Only A:", sA - sB)
print("Unique:", sA | sB)
```

## Exercise 6: Nested Dict

```python
students = {"Alice": {"subjects": ["Math", "Physics"]}, "Bob": {"subjects": ["CS", "Stats"]}, "Charlie": {"subjects": ["ML", "Data Mining"]}}
charlie = students.get("Charlie", {})
subs = charlie.get("subjects", [])
second = subs[1] if len(subs) > 1 else None
print("Second subject:", second)
```
