## Variables - Labels, Not Boxes

In Python, variables are labels on objects, not boxes for values.

```python
x = 42
y = x
print("x=", x, "y=", y)
print("x is y:", x is y)

x = "hello"
print("Now x=", x, "but y still=", y)
```

## Core Types

```python
age = 25
pi = 3.14159
name = "Data Science"
flag = True
nothing = None
data = [1, 2, 3]
coords = (10.5, 20.3)
person = {"name": "Alice", "age": 30}
ids = {101, 102}

for var, val in [("age",age),("pi",pi),("name",name),("flag",flag),("nothing",nothing),("data",data),("coords",coords),("person",person),("ids",ids)]:
    print(f"{var:>8} -> {type(val).__name__:>6} = {repr(val)}")
```

## Float Precision Warning

```python
# Floating point is not exact!
print("0.1 + 0.2 =", 0.1 + 0.2)
print("Equal to 0.3?", 0.1 + 0.2 == 0.3)

from math import isclose
print("Using isclose:", isclose(0.1 + 0.2, 0.3))

# Always use isclose() for float comparisons
```

## Strings and f-strings

```python
# String operations
csv = "John,25,Engineer"
print("Split:", csv.split(","))

# f-strings
name = "Alice"
age = 30
sal = 75000.50
print(f"Name: {name}, Age: {age}")
print(f"Salary: ${sal:,.2f}")
print(f"Percent: {0.856:.1%}")

# Common string methods
print("  hello  ".strip())
print("HELLO".lower())
print("hello world".title())
```

## Lists

```python
prices = [12.5, 15.3, 9.8, 22.1, 18.6]
print("Prices:", prices)
print("Max:", max(prices), "Min:", min(prices))
print("Mean:", sum(prices)/len(prices))

# Slicing
print("prices[1:3]:", prices[1:3])
print("prices[-2:]:", prices[-2:])
print("Every other:", prices[::2])

# Copy trap
orig = [1, 2, 3]
bad = orig
bad.append(4)
print("Original after bad.append:", orig)

good = orig.copy()
good.append(5)
print("Original:", orig, "Copy:", good)
```

## Dicts and Sets

```python
# Dicts
student = {"name": "Alice", "grade": 92, "subjects": ["Math", "ML"]}
print("Grade:", student["grade"])
print("City:", student.get("city", "Unknown"))

# Dict comprehension
squares = {x: x**2 for x in range(5)}
print("Squares:", squares)

# Sets
a = {1,2,3,4}
b = {3,4,5,6}
print("Union:", a|b)
print("Intersection:", a&b)
print("Difference:", a-b)
```
