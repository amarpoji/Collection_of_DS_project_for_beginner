## Setup: Connect to the Database

Run the cell below to connect to `sql_mastery.db`.

```python
import sqlite3
import pandas as pd

DB_PATH = r'C:\Users\USER\Desktop\agentic ai\notebook_ds\sql_mastery_course\sql_mastery.db'
conn = sqlite3.connect(DB_PATH)
print(f"Connected to: {DB_PATH}")

def run_query(query):
    return pd.read_sql_query(query, conn)

# Test the connection
run_query("SELECT name FROM sqlite_master WHERE type='table' ORDER BY name;")
```

## Explore the Database

```python
# See schema of employees
run_query("PRAGMA table_info(employees);")
```

## SELECT & WHERE

```python
# Basic SELECT
run_query("SELECT first_name, last_name, salary FROM employees LIMIT 10;")
```

```python
# WHERE filtering
run_query("SELECT * FROM employees WHERE salary >= 100000 ORDER BY salary DESC;")
```

```python
# Multiple conditions
run_query("""SELECT first_name, last_name, job_title, salary FROM employees WHERE department_id = 1 AND salary > 80000 ORDER BY salary DESC;""")
```

```python
# LIKE wildcard
run_query("SELECT * FROM customers WHERE city LIKE 'N%';")
```

## ORDER BY, LIMIT, DISTINCT

```python
# Top 5 highest paid
run_query("SELECT first_name, last_name, salary FROM employees ORDER BY salary DESC LIMIT 5;")
```

```python
# DISTINCT departments
run_query("SELECT DISTINCT department_id FROM employees ORDER BY department_id;")
```

## Aggregate Functions & GROUP BY

```python
run_query("""SELECT COUNT(*) AS total_employees, AVG(salary) AS avg_salary, MIN(salary) AS min_salary, MAX(salary) AS max_salary FROM employees;""")
```

```python
# Average salary by department
run_query("""SELECT department_id, COUNT(*) AS emp_count, ROUND(AVG(salary), 0) AS avg_salary FROM employees GROUP BY department_id ORDER BY avg_salary DESC;""")
```

```python
# HAVING: depts with avg salary > 80k
run_query("""SELECT department_id, COUNT(*) AS emp_count, ROUND(AVG(salary), 0) AS avg_salary FROM employees GROUP BY department_id HAVING AVG(salary) > 80000 ORDER BY avg_salary DESC;""")
```
