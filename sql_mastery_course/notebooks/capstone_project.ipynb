```python
import sqlite3
import pandas as pd
conn = sqlite3.connect('..\sql_mastery.db')
def sql(q): return pd.read_sql_query(q, conn)
print("Ready!")
```

## Section A: Revenue Analysis

```python
# Monthly Revenue Trend
sql("""SELECT STRFTIME('%Y-%m', order_date) AS month, COUNT(*) AS orders, ROUND(SUM(total_amount), 2) AS revenue FROM orders WHERE status = 'Delivered' GROUP BY month ORDER BY month;""")
```

```python
# Category Revenue
sql("""SELECT p.category, COUNT(oi.quantity) AS units_sold, ROUND(SUM(oi.quantity * oi.unit_price), 2) AS revenue FROM order_items oi JOIN products p ON oi.product_id = p.product_id GROUP BY p.category ORDER BY revenue DESC;""")
```

## Section B: Customer Analysis

```python
# Customer Lifetime Value
sql("""SELECT c.first_name || ' ' || c.last_name AS name, COUNT(DISTINCT o.order_id) AS orders, ROUND(SUM(o.total_amount), 2) AS total_spent FROM customers c JOIN orders o ON c.customer_id = o.customer_id GROUP BY c.customer_id ORDER BY total_spent DESC;""")
```

## Section C: Operations

```python
# Order Status
sql("SELECT status, COUNT(*) AS count, ROUND(COUNT(*)*100.0/(SELECT COUNT(*) FROM orders),1) AS pct FROM orders GROUP BY status;")
```

```python
# Shipping by State
sql("""SELECT shipping_state, COUNT(*) AS orders, ROUND(SUM(total_amount),2) AS revenue FROM orders GROUP BY shipping_state ORDER BY revenue DESC;""")
```

## Section D: Executive KPI Summary

```python
sql("""SELECT (SELECT ROUND(SUM(total_amount),2) FROM orders WHERE status='Delivered') AS total_revenue, (SELECT COUNT(*) FROM orders) AS total_orders, (SELECT ROUND(AVG(total_amount),2) FROM orders WHERE status='Delivered') AS avg_order_value, (SELECT COUNT(*) FROM customers) AS total_customers, (SELECT COUNT(DISTINCT customer_id) FROM orders) AS active_customers;""")
```
