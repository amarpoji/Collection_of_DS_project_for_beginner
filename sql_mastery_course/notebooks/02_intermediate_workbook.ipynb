# SQL Mastery Course - Intermediate Level

JOINs, Subqueries, CTEs, CASE WHEN, Window Functions.

```python
import sqlite3
import pandas as pd
conn = sqlite3.connect('..\sql_mastery.db')
def run_query(q): return pd.read_sql_query(q, conn)
print("Connected!")
```

## JOINs

```python
# INNER JOIN: employees + departments
run_query("""SELECT e.first_name, e.last_name, e.salary, d.department_name FROM employees e INNER JOIN departments d ON e.department_id = d.department_id ORDER BY e.salary DESC;""")
```

```python
# LEFT JOIN: all customers with orders
run_query("""SELECT c.first_name, c.last_name, o.order_id, o.total_amount, o.status FROM customers c LEFT JOIN orders o ON c.customer_id = o.customer_id ORDER BY c.last_name;""")
```

```python
# Multi-table JOIN
run_query("""SELECT o.order_id, c.first_name || ' ' || c.last_name AS customer, p.product_name, oi.quantity FROM orders o JOIN customers c ON o.customer_id = c.customer_id JOIN order_items oi ON o.order_id = oi.order_id JOIN products p ON oi.product_id = p.product_id ORDER BY o.order_id LIMIT 15;""")
```

```python
# Self-JOIN: employees + managers
run_query("""SELECT e.first_name || ' ' || e.last_name AS employee, e.job_title, m.first_name || ' ' || m.last_name AS manager FROM employees e LEFT JOIN employees m ON e.manager_id = m.employee_id;""")
```

## Subqueries

```python
# Above average salary
run_query("SELECT first_name, last_name, salary FROM employees WHERE salary > (SELECT AVG(salary) FROM employees) ORDER BY salary DESC;")
```

```python
# Subquery in SELECT
run_query("""SELECT e.first_name, e.last_name, e.salary, (SELECT ROUND(AVG(salary)) FROM employees WHERE department_id = e.department_id) AS dept_avg FROM employees e ORDER BY e.department_id;""")
```

## CASE WHEN

```python
# Salary brackets
run_query("""SELECT first_name, last_name, salary, CASE WHEN salary < 70000 THEN 'Junior' WHEN salary BETWEEN 70000 AND 100000 THEN 'Mid' WHEN salary BETWEEN 100001 AND 150000 THEN 'Senior' ELSE 'Executive' END AS level FROM employees ORDER BY salary;""")
```

## CTEs

```python
# Top products by revenue
run_query("""WITH product_revenue AS (SELECT p.product_id, p.product_name, SUM(oi.quantity * oi.unit_price) AS revenue FROM order_items oi JOIN products p ON oi.product_id = p.product_id GROUP BY p.product_id) SELECT * FROM product_revenue ORDER BY revenue DESC LIMIT 5;""")
```

## Window Functions

```python
# Rank by salary within department
run_query("""SELECT first_name, last_name, department_id, salary, ROW_NUMBER() OVER (PARTITION BY department_id ORDER BY salary DESC) AS sal_rank FROM employees ORDER BY department_id, sal_rank;""")
```

```python
# Running total of orders
run_query("SELECT order_id, order_date, total_amount, SUM(total_amount) OVER (ORDER BY order_date) AS running_total FROM orders ORDER BY order_date;")
```
